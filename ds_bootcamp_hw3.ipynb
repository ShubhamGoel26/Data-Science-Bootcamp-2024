```python
df['time_of_day'] = df['hour_beginning'].dt.time
```

```python
df
```

```python
def custom(hour):
    if hour >= 4 and hour < 12:
        return 'Morning'
    elif hour >= 12 and hour < 17:
        return 'Afternoon'
    elif hour >= 17 and hour < 21:
        return 'Evening'
    else:
        return 'Night'
```

```python
df['time_of_day'] = df['time_of_day'].apply(lambda x: custom(x.hour))
```
